# RAG Evaluation

```python
questions = [
    "Why are customers unhappy with credit cards?",
    "What problems do customers report with personal loans?",
    "What complaints are common about savings accounts?",
    "Why do customers complain about money transfers?",
    "Which companies receive the most complaints about credit cards?",
    "What customer service issues appear most frequently?",
    "What fraud-related complaints are common?",
    "What billing problems do customers report?"
]
```

```python
for question in questions:

    print("=" * 100)
    print("QUESTION:")
    print(question)

    answer, sources = ask_rag(question)

    print("\nANSWER:")
    print(answer)

    print("\nTOP SOURCE:")

    print(sources[0]["metadata"])

    print(sources[0]["text"][:400])

    print("\n")
```

| Question | Quality (1-5) | Comments |
|----------|---------------|----------|
| Why are customers unhappy with credit cards? | 5 | Answer matches retrieved complaints well. |
| What problems do customers report with personal loans? | 4 | Mostly accurate, but could summarize better. |
| What complaints are common about savings accounts? | 5 | Retrieved relevant complaint excerpts. |
| Why do customers complain about money transfers? | 4 | Good retrieval, slightly repetitive answer. |
| Which companies receive the most complaints? | 3 | Retrieved correct companies but lacked aggregation. |
| What customer service issues appear most frequently? | 5 | Strong evidence from retrieved chunks. |
| What fraud-related complaints are common? | 5 | Correctly grounded in complaint narratives. |
| What billing problems do customers report? | 4 | Good answer, but some repeated phrases. |
